# NB_Carriers: energy per kilogram and per liter (Fig. 2)
Every value is typed in the first cell with its source. Nothing is estimated except one derived point, iron powder poured, which is the mean measured poured density divided by solid iron's density, times solid iron's value. Run all cells.

```python
import numpy as np
# name: (MJ/kg, MJ/L, family, source)
DATA = {
 'Al':(31.05,83.85,'metal','Fischer & Grubelich, Sandia 1998 (heat of combustion to oxide)'),
 'Si':(32.43,80.74,'metal','same'), 'Ti':(19.72,88.74,'metal','same'), 'Mg':(26.11,45.44,'metal','same'),
 'Zn':(5.36,38.27,'metal','same'), 'Li':(45.86,24.49,'metal','same'),
 'iron, solid':(7.36,57.92,'iron','Janicka 2023, Table 2'),
 'gasoline':(43.448,32.36,'fossil','Argonne GREET 1.8d.1 heating values (LHV)'),
 'diesel':(42.612,36.09,'fossil','same'), 'LNG':(48.632,20.83,'fossil','same'),
 'liquid hydrogen':(120.07,8.49,'H2/NH3','Aziz 2020'), 'compressed hydrogen (69 MPa)':(120.0,4.5,'H2/NH3','Aziz 2020'),
 'liquid ammonia':(18.6,11.2,'H2/NH3','Aziz 2020'),
 'Li-ion cell':(0.97,2.34,'battery','Frith 2023'), 'Li-ion pack':(0.49,0.76,'battery','Frith 2023'),
}
COAL = dict(kg=(25,36), L=(30,57), source='Janicka 2023, Table 2 (bituminous coal)')
poured = [3.44,3.55,3.35]; solid = 7.88      # g/cm3, van Rooij thesis, Table 4.2
frac = np.mean(poured)/solid
iron_powder_MJL = DATA['iron, solid'][1]*frac
print(f'poured/solid density = {frac:.3f}  ->  iron powder poured = {iron_powder_MJL:.1f} MJ/L')
for k,v in DATA.items(): print(f'{k:32s} {v[0]:8.2f} MJ/kg {v[1]:7.2f} MJ/L  [{v[3]}]')
assert abs(frac-0.44)<0.01 and abs(iron_powder_MJL-25.4)<0.5    # the paper: 43 to 45 % of solid, about 25 MJ/L
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse, Rectangle
col={'metal':'#0077BB','fossil':'#444444','H2/NH3':'#EE7733','battery':'#AA3377','iron':'#111111'}
fig,ax=plt.subplots(figsize=(6.5,3.35))
for k,(kg,L,fam,_) in DATA.items(): ax.loglog(kg,L,'s' if fam in('metal','battery','iron') else 'o',color=col[fam],ms=8 if fam=='iron' else 5)
ax.add_patch(Rectangle((COAL['kg'][0],COAL['L'][0]),COAL['kg'][1]-COAL['kg'][0],COAL['L'][1]-COAL['L'][0],fc='#bbb',alpha=.55,zorder=0))
ax.annotate('',xy=(7.36,iron_powder_MJL),xytext=(7.36,57.92),arrowprops=dict(arrowstyle='-|>'))
ax.plot(7.36,iron_powder_MJL,'s',mfc='white',mec='k',ms=7)
ax.text(6.5,57.9,'iron, solid',ha='right',va='center',fontweight='bold'); ax.text(6.5,iron_powder_MJL,'iron powder, poured',ha='right',va='center'); ax.text(23,36,'coal',ha='right',fontweight='bold')
ax.set_xlabel('Energy per kilogram (MJ/kg)'); ax.set_ylabel('Energy per liter (MJ/L)'); ax.grid(alpha=.25)
fig.savefig('fig_carriers.png',dpi=200,bbox_inches='tight'); plt.show()
```

**Reading the figure.** Iron holds less per kilogram than coal (7.36 against 25 to 36 MJ/kg). Per liter, solid iron is slightly above coal's range, and poured powder (about 25 MJ/L) is below it. The case for iron is that it is a solid that can be recharged with hydrogen; this figure does not show an efficiency advantage and no efficiency is plotted here.
